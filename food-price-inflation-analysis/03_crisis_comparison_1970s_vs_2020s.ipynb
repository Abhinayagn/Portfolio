```python
# 03_crisis_comparison_1970s_vs_2020s.ipynb

import pandas as pd

print("Libraries loaded successfully.")

# ------------------------------------------------------------
# 1. Load the cleaned inflation data from notebook 2
# ------------------------------------------------------------

df = pd.read_csv("abhinaya_cleaned_food_inflation.csv")

print("Data loaded.")
print("Shape:", df.shape)
display(df.head(3))

# Basic checks
required_cols = ["Year", "Decade", "Percent change", "Consumer Price Index item"]
missing = [c for c in required_cols if c not in df.columns]
if missing:
    raise ValueError(f"Missing required columns: {missing}")

# ------------------------------------------------------------
# 2. Split into 1970s and 2020s
# ------------------------------------------------------------

data_1970s = df[df["Decade"] == 1970]
data_2020s = df[df["Decade"] == 2020]

print("1970s records:", len(data_1970s))
print("2020s records:", len(data_2020s))

# ------------------------------------------------------------
# 3. Summary statistics for each decade
# ------------------------------------------------------------

def summarize_decade(data, label):
    print(f"\n{label} STATISTICS")
    print("=" * 50)
    print(f"Average Inflation: {data['Percent change'].mean():.2f}%")
    print(f"Volatility (Std Dev): {data['Percent change'].std():.2f}%")
    print(f"Worst Spike: {data['Percent change'].max():.2f}%")
    print(f"Worst Drop: {data['Percent change'].min():.2f}%")
    print(f"Data Points: {len(data)}")

print("\n1970s crisis statistics:")
summarize_decade(data_1970s, "1970s")

print("\n2020s current period statistics:")
summarize_decade(data_2020s, "2020s")

# ------------------------------------------------------------
# 4. Direct numerical comparison
# ------------------------------------------------------------

avg_1970s = data_1970s["Percent change"].mean()
avg_2020s = data_2020s["Percent change"].mean()
vol_1970s = data_1970s["Percent change"].std()
vol_2020s = data_2020s["Percent change"].std()

print("\nCOMPARISON (2020s vs 1970s)")
print("=" * 50)

diff_avg = avg_2020s - avg_1970s
diff_vol = vol_2020s - vol_1970s

if diff_avg < 0:
    pct_less_severe = abs(diff_avg / avg_1970s * 100)
    print(f"2020s inflation is {abs(diff_avg):.2f} percentage points LOWER than 1970s.")
    print(f"Current crisis is approximately {pct_less_severe:.1f}% less severe than the 1970s crisis.")
else:
    print(f"2020s inflation is {diff_avg:.2f} percentage points HIGHER than 1970s.")

if diff_vol < 0:
    pct_less_volatile = abs(diff_vol / vol_1970s * 100)
    print(f"2020s are {abs(diff_vol):.2f} percentage points LESS volatile.")
    print(f"That is about {pct_less_volatile:.1f}% lower volatility than the 1970s.")
else:
    print(f"2020s are {diff_vol:.2f} percentage points MORE volatile.")

# ------------------------------------------------------------
# 5. Foods driving each crisis
# ------------------------------------------------------------

print("\nFOODS DRIVING EACH CRISIS")
print("=" * 50)

print("\n1970s - Top 5 highest inflation foods (average over decade):")
foods_1970s = (
    data_1970s.groupby("Consumer Price Index item")["Percent change"]
              .mean()
              .sort_values(ascending=False)
              .head(5)
)
for food, val in foods_1970s.items():
    print(f"- {food}: {val:.2f}% average inflation")

print("\n2020s - Top 5 highest inflation foods (average over decade):")
foods_2020s = (
    data_2020s.groupby("Consumer Price Index item")["Percent change"]
              .mean()
              .sort_values(ascending=False)
              .head(5)
)
for food, val in foods_2020s.items():
    print(f"- {food}: {val:.2f}% average inflation")

# ------------------------------------------------------------
# 6. Save comparison outputs
# ------------------------------------------------------------

comparison_data = pd.DataFrame({
    "Decade": ["1970s", "2020s"],
    "Avg_Inflation": [avg_1970s, avg_2020s],
    "Volatility": [vol_1970s, vol_2020s],
    "Max_Spike": [data_1970s["Percent change"].max(), data_2020s["Percent change"].max()],
    "Max_Drop": [data_1970s["Percent change"].min(), data_2020s["Percent change"].min()],
})

comparison_data.to_csv("abhinaya_1970s_vs_2020s.csv", index=False)

print("\nSaved file:")
print("- abhinaya_1970s_vs_2020s.csv")
```
