```python
# 01_data_loading_and_profiling.ipynb

# If needed (Colab/new env), run once:
# !pip install pandas openpyxl matplotlib numpy jupyter

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries loaded successfully!")
print(f"Pandas version: {pd.__version__}")

# ------------------------------------------------------------------
# 1. Load raw data (update paths if needed)
# ------------------------------------------------------------------

cpi_forecast_path = r"C:\Users\abhin\OneDrive\Desktop\food-price-inflation-analysis\DATA\Raw\CPIForecast.csv"
historical_cpi_path = r"C:\Users\abhin\OneDrive\Desktop\food-price-inflation-analysis\DATA\Raw\historicalcpi.csv"

df_forecast = pd.read_csv(cpi_forecast_path)
df_cpi = pd.read_csv(historical_cpi_path)

print("Loaded successfully.")
print("CPI Forecast shape:", df_forecast.shape)
print("Historical CPI shape:", df_cpi.shape)

display(df_forecast.head(3))
display(df_cpi.head(3))

# ------------------------------------------------------------------
# 2. Profile CPI Forecast data
# ------------------------------------------------------------------

print("\n=== CPI FORECAST PROFILING ===")
print(df_forecast.info())
print("\nMissing values:")
print(df_forecast.isna().sum())
print("\nSummary statistics:")
display(df_forecast.describe(include="all").transpose())
print("\nSample rows:")
display(df_forecast.head(10))

# ------------------------------------------------------------------
# 3. Profile Historical CPI data
# ------------------------------------------------------------------

print("\n=== HISTORICAL CPI PROFILING ===")
print(df_cpi.info())
print("\nMissing values:")
print(df_cpi.isna().sum())
print("\nSummary statistics:")
display(df_cpi.describe(include="all").transpose())

print("\nTop 10 CPI items by count:")
print(df_cpi["Consumer Price Index item"].value_counts().head(10))

print("\nUnique food items (first 20):")
print(df_cpi["Consumer Price Index item"].unique()[:20])
print(f"\nTotal distinct CPI items: {df_cpi['Consumer Price Index item'].nunique()}")

print("\nAll column names:")
print(df_cpi.columns.tolist())

# ------------------------------------------------------------------
# 4. Save an initial cleaned copy (no transformations yet)
# ------------------------------------------------------------------

df_cpi.to_csv("cleaned_cpi_data_initial.csv", index=False)
print("\nSaved: cleaned_cpi_data_initial.csv")
```
