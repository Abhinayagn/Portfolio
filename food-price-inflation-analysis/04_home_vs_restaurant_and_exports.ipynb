```python
# 04_home_vs_restaurant_and_exports.ipynb

import pandas as pd

print("Libraries loaded successfully.")

# ------------------------------------------------------------
# 1. Load cleaned inflation data
# ------------------------------------------------------------

df = pd.read_csv("abhinaya_cleaned_food_inflation.csv")

print("Data loaded.")
print("Shape:", df.shape)
display(df.head(3))

required_cols = ["Year", "Decade", "Percent change", "Consumer Price Index item"]
missing = [c for c in required_cols if c not in df.columns]
if missing:
    raise ValueError(f"Missing required columns: {missing}")

# ------------------------------------------------------------
# 2. Home vs restaurant comparison (if categories exist)
# ------------------------------------------------------------

print("=" * 70)
print("HOME VS RESTAURANT COMPARISON")
print("=" * 70)

food_at_home = df[df["Consumer Price Index item"]
                  .str.contains("at home", case=False, na=False)]
food_away = df[df["Consumer Price Index item"]
               .str.contains("away from home", case=False, na=False)]

if len(food_at_home) > 0 and len(food_away) > 0:
    print("\nFound 'Food at home' and 'Food away from home' categories.")

    print("\nFOOD AT HOME (Groceries):")
    print(f"  Average inflation (all years): {food_at_home['Percent change'].mean():.2f}%")
    print(f"  Volatility: {food_at_home['Percent change'].std():.2f}%")
    print(f"  2020s average: {food_at_home[food_at_home['Decade'] == 2020]['Percent change'].mean():.2f}%")

    print("\nFOOD AWAY FROM HOME (Restaurants):")
    print(f"  Average inflation (all years): {food_away['Percent change'].mean():.2f}%")
    print(f"  Volatility: {food_away['Percent change'].std():.2f}%")
    print(f"  2020s average: {food_away[food_away['Decade'] == 2020]['Percent change'].mean():.2f}%")

    home_vs_away = pd.DataFrame({
        "Category": ["Food at Home", "Food Away from Home"],
        "Avg_Inflation_All_Years": [
            food_at_home["Percent change"].mean(),
            food_away["Percent change"].mean()
        ],
        "Volatility_All_Years": [
            food_at_home["Percent change"].std(),
            food_away["Percent change"].std()
        ],
        "Avg_Inflation_2020s": [
            food_at_home[food_at_home["Decade"] == 2020]["Percent change"].mean(),
            food_away[food_away["Decade"] == 2020]["Percent change"].mean()
        ]
    })

    home_vs_away.to_csv("home_vs_restaurant_comparison.csv", index=False)
    print("\nSaved: home_vs_restaurant_comparison.csv")

else:
    print("\nNo 'Food at home' and 'Food away from home' categories found.")
    print("Data appears to have item-level categories only (eggs, beef, etc.).")
    print("\nShowing overall inflation instead:")
    print(f"Overall average inflation: {df['Percent change'].mean():.2f}%")
    print(f"Overall volatility: {df['Percent change'].std():.2f}%")

# ------------------------------------------------------------
# 3. Final cleaned export for sharing
# ------------------------------------------------------------

df.to_csv("abhinaya_cleaned_data.csv", index=False)
print("\nSaved: abhinaya_cleaned_data.csv")

print("\nNotebook 4 complete.")
```
